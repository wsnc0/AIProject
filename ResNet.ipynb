```python
# Skin Disease Classification using ResNet

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import shutil
from tqdm import tqdm
import glob

# Pytorch imports
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision
from torchvision import transforms, models
from torchvision.datasets import ImageFolder
from torchvision.utils import make_grid
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
# Sklearn metrics
from sklearn.metrics import confusion_matrix, classification_report, f1_score, precision_score, recall_score

# Set random seeds for reproducibility
torch.manual_seed(42)
np.random.seed(42)

# Force CPU usage due to CUDA kernel compatibility issues
device = torch.device('cpu')
print(f"Using device: {device}")

# Define directories
processed_data_dir = './processedData'
model_logs_dir = './modelLogs'
os.makedirs(model_logs_dir, exist_ok=True)

# Define data transformations
data_transforms = {
    'train': transforms.Compose([
        transforms.RandomHorizontalFlip(),
        transforms.RandomVerticalFlip(),
        transforms.RandomRotation(10),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
    'val': transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
    'test': transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
}

# Load datasets
print("Loading datasets...")
image_datasets = {
    split: ImageFolder(
        os.path.join(processed_data_dir, split), 
        data_transforms[split]
    )
    for split in ['train', 'val', 'test']
}

# Create dataloaders - reduced batch size and workers for CPU
dataloaders = {
    split: DataLoader(
        image_datasets[split], 
        batch_size=16,  # Reduced batch size 
        shuffle=(split == 'train'), 
        num_workers=0   # No multiprocessing for stability
    )
    for split in ['train', 'val', 'test']
}

dataset_sizes = {split: len(image_datasets[split]) for split in ['train', 'val', 'test']}
class_names = image_datasets['train'].classes
num_classes = len(class_names)

print(f"Number of classes: {num_classes}")
print(f"Class names: {class_names}")
for split in ['train', 'val', 'test']:
    print(f"{split.capitalize()} set size: {dataset_sizes[split]}")

# Function to display a batch of images
def imshow(inp, title=None):
    inp = inp.cpu().numpy().transpose((1, 2, 0))
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    inp = std * inp + mean
    inp = np.clip(inp, 0, 1)
    plt.figure(figsize=(15, 15))
    plt.imshow(inp)
    if title is not None:
        plt.title(title)
    plt.pause(0.001)

# Show sample images
try:
    inputs, classes = next(iter(dataloaders['train']))
    out = make_grid(inputs[:4])  # Showing fewer samples to save memory
    class_labels = [class_names[x] for x in classes[:4]]
    imshow(out, title=str(class_labels))
except Exception as e:
    print(f"Error displaying sample images: {e}")

# Define a simpler ResNet model
def create_model(num_classes):
    # Use ResNet18 instead of ResNet50 for faster CPU training
    model = models.resnet18(weights=None)  # No pretrained weights to avoid CUDA issues
    
    # Replace final fully connected layer
    num_ftrs = model.fc.in_features
    model.fc = nn.Linear(num_ftrs, num_classes)
    
    return model

# Initialize model
model = create_model(num_classes)
model = model.to(device)

# Loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.1, patience=5, verbose=True)

# Training function
def train_model(model, criterion, optimizer, scheduler, num_epochs=10):  # Reduced epochs
    best_model_wts = model.state_dict()
    best_acc = 0.0
    
    # For tracking metrics
    history = {
        'train_loss': [], 'train_acc': [],
        'val_loss': [], 'val_acc': []
    }
    
    for epoch in range(num_epochs):
        print(f'Epoch {epoch+1}/{num_epochs}')
        print('-' * 10)
        
        # Each epoch has a training and validation phase
        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()
            else:
                model.eval()
            
            running_loss = 0.0
            running_corrects = 0
            
            # Iterate over data
            for inputs, labels in tqdm(dataloaders[phase], desc=f"{phase} batches"):
                inputs = inputs.to(device)
                labels = labels.to(device)
                
                # Zero the parameter gradients
                optimizer.zero_grad()
                
                # Forward pass
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)
                    
                    # Backward pass + optimize only if training
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()
                
                # Statistics
                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)
            
            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]
            
            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')
            
            # Save metrics
            if phase == 'train':
                history['train_loss'].append(epoch_loss)
                history['train_acc'].append(epoch_acc.item())
            else:
                history['val_loss'].append(epoch_loss)
                history['val_acc'].append(epoch_acc.item())
                # Update scheduler based on validation loss
                scheduler.step(epoch_loss)
            
            # Save best model
            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = model.state_dict().copy()
                
                # Save best model checkpoint
                torch.save({
                    'epoch': epoch,
                    'model_state_dict': best_model_wts,
                    'optimizer_state_dict': optimizer.state_dict(),
                    'accuracy': best_acc,
                }, os.path.join(model_logs_dir, 'best_model.pth'))
        
        # Save checkpoint for each epoch
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'history': history,
        }, os.path.join(model_logs_dir, f'checkpoint_epoch_{epoch+1}.pth'))
        
        print()
    
    # Load best model weights
    model.load_state_dict(best_model_wts)
    return model, history

# Train the model
print("Starting model training...")
model, history = train_model(model, criterion, optimizer, scheduler, num_epochs=10)  # Reduced epochs for CPU
print("Training complete!")

# Plot training history
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history['train_loss'], label='Train Loss')
plt.plot(history['val_loss'], label='Validation Loss')
plt.title('Loss vs. Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history['train_acc'], label='Train Accuracy')
plt.plot(history['val_acc'], label='Validation Accuracy')
plt.title('Accuracy vs. Epochs')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.tight_layout()
plt.savefig(os.path.join(model_logs_dir, 'training_history.png'))
plt.show()

# Evaluate model on test set
def evaluate_model(model, dataloader):
    model.eval()
    
    all_preds = []
    all_labels = []
    
    with torch.no_grad():
        for inputs, labels in tqdm(dataloader, desc="Evaluating"):
            inputs = inputs.to(device)
            labels = labels.to(device)
            
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    
    return all_preds, all_labels

# Run evaluation
print("Evaluating model on test set...")
y_pred, y_true = evaluate_model(model, dataloaders['test'])
print("Evaluation complete!")

# Calculate metrics
accuracy = np.mean(np.array(y_pred) == np.array(y_true))
f1 = f1_score(y_true, y_pred, average='weighted')
precision = precision_score(y_true, y_pred, average='weighted')
recall = recall_score(y_true, y_pred, average='weighted')

print(f"Test Accuracy: {accuracy:.4f}")
print(f"F1 Score (weighted): {f1:.4f}")
print(f"Precision (weighted): {precision:.4f}")
print(f"Recall (weighted): {recall:.4f}")

# Generate classification report
report = classification_report(y_true, y_pred, target_names=class_names, output_dict=True)
report_df = pd.DataFrame(report).transpose()
print("Classification Report:")
print(report_df)

# Save report to CSV
report_df.to_csv(os.path.join(model_logs_dir, 'classification_report.csv'))

# Plot confusion matrix
plt.figure(figsize=(12, 10))
cm = confusion_matrix(y_true, y_pred)
cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('Normalized Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig(os.path.join(model_logs_dir, 'confusion_matrix.png'))
plt.show()

# Save final model
torch.save({
    'model_state_dict': model.state_dict(),
    'class_names': class_names,
    'metrics': {
        'accuracy': accuracy,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }
}, os.path.join(model_logs_dir, 'final_model.pth'))

print("Model saved! Evaluation metrics and visualizations have been saved to the modelLogs directory.")

# Function to predict on new images
def predict_image(image_path, model, class_names):
    from PIL import Image
    
    # Define the same transformation as used for test set
    transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
    
    # Load and transform the image
    image = Image.open(image_path).convert('RGB')
    image_tensor = transform(image).unsqueeze(0).to(device)
    
    # Make prediction
    model.eval()
    with torch.no_grad():
        outputs = model(image_tensor)
        _, predicted = torch.max(outputs, 1)
        
        # Get probabilities
        probabilities = torch.nn.functional.softmax(outputs, dim=1)[0]
        
    predicted_class = class_names[predicted.item()]
    probability = probabilities[predicted].item()
    
    return predicted_class, probability, probabilities.cpu().numpy()

# Example usage of prediction function (uncomment to use)
"""
image_path = 'path/to/your/image.jpg'
predicted_class, probability, all_probs = predict_image(image_path, model, class_names)
print(f"Predicted class: {predicted_class} with probability: {probability:.4f}")

# Display the image with prediction
plt.figure(figsize=(8, 8))
image = plt.imread(image_path)
plt.imshow(image)
plt.title(f"Prediction: {predicted_class} ({probability:.2f})")
plt.axis('off')
plt.show()

# Show top 3 predictions
top_3_indices = np.argsort(all_probs)[-3:][::-1]
top_3_classes = [class_names[i] for i in top_3_indices]
top_3_probs = all_probs[top_3_indices]

plt.figure(figsize=(10, 5))
plt.barh(top_3_classes, top_3_probs)
plt.xlabel('Probability')
plt.title('Top 3 Predictions')
plt.xlim(0, 1)
plt.tight_layout()
plt.show()
"""
```

```python
# Imports remain the same
from sklearn.metrics import confusion_matrix, classification_report, f1_score, precision_score, recall_score
import os
import torch
import numpy as np
from torchvision import transforms, models
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader, Dataset
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from tqdm import tqdm
import seaborn as sns
import pandas as pd
from PIL import Image

# Set seeds
torch.manual_seed(42)
np.random.seed(42)
device = torch.device('cpu')
print(f"Using device: {device}")

# Directories
processed_data_dir = './processedData'
model_logs_dir = './modelLogs'
os.makedirs(model_logs_dir, exist_ok=True)

# Define class-specific transforms
light_transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize
    transforms.Pad(padding=4),      # Padding
    transforms.RandomHorizontalFlip(),  # Flipping
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])  # Normalize
])

heavy_transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize
    transforms.Pad(padding=4),      # Padding
    transforms.RandomHorizontalFlip(),  # Flipping
    transforms.RandomVerticalFlip(),    # Flipping
    transforms.RandomRotation(15),      # Rotation
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3),  # Color jitter
    transforms.RandomAffine(degrees=0, translate=(0.1, 0.1)),  # Affine transformation
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])  # Normalize
])

# Validation and test transforms (simple)
val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize
    transforms.Pad(padding=4),      # Padding
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])  # Normalize
])

# Custom dataset for class-specific transforms
class SkinLesionDataset(Dataset):
    def __init__(self, root_dir, split, light_classes=None):
        self.dataset = ImageFolder(os.path.join(root_dir, split))
        self.split = split
        # If no light_classes are specified, default to "melanoma" and "melanocytic nevi"
        self.light_classes = light_classes or ["melanoma", "melanocytic nevi"]
        self.light_class_indices = [self.dataset.class_to_idx[cls] for cls in self.light_classes if cls in self.dataset.class_to_idx]
        
        print(f"Light augmentation applied to classes: {self.light_classes}")
        print(f"Light augmentation class indices: {self.light_class_indices}")
        
    def __len__(self):
        return len(self.dataset)
    
    def __getitem__(self, idx):
        image_path, label = self.dataset.samples[idx]
        image = Image.open(image_path).convert('RGB')
        
        # Apply transforms based on class and split
        if self.split == 'train':
            if label in self.light_class_indices:
                # Light augmentation for melanoma and melanocytic nevi
                transform = light_transform
            else:
                # Heavy augmentation for other classes
                transform = heavy_transform
        else:
            # Simple transforms for validation and test
            transform = val_test_transform
            
        image = transform(image)
        return image, label

# Load datasets with class-specific augmentation
print("Loading datasets with class-specific augmentation...")
datasets = {
    split: SkinLesionDataset(processed_data_dir, split, ["melanoma", "melanocytic nevi"])
    for split in ['train', 'val', 'test']
}

# Dataloaders
dataloaders = {
    split: DataLoader(datasets[split], batch_size=16, shuffle=(split == 'train'), num_workers=0)
    for split in ['train', 'val', 'test']
}

# Get class names (need to use the underlying ImageFolder dataset)
class_names = datasets['train'].dataset.classes
num_classes = len(class_names)
dataset_sizes = {split: len(datasets[split]) for split in ['train', 'val', 'test']}

print(f"Number of classes: {num_classes}")
print(f"Class names: {class_names}")
for split in ['train', 'val', 'test']:
    print(f"{split.capitalize()} set size: {dataset_sizes[split]}")

# Display sample images
def imshow(inp, title=None):
    inp = inp.cpu().numpy().transpose((1, 2, 0))
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    inp = std * inp + mean
    inp = np.clip(inp, 0, 1)
    plt.figure(figsize=(15, 15))
    plt.imshow(inp)
    if title is not None:
        plt.title(title)
    plt.pause(0.001)

try:
    inputs, classes = next(iter(dataloaders['train']))
    out = make_grid(inputs[:4])
    # Need to get class names mapped to indices
    class_labels = [datasets['train'].dataset.classes[x] for x in classes[:4]]
    imshow(out, title=str(class_labels))
except Exception as e:
    print(f"Error displaying sample images: {e}")

# Modified model function to freeze all layers except the last two
def create_model(num_classes):
    model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)  # Use pretrained
    
    # Freeze all layers in the base model
    for param in model.parameters():
        param.requires_grad = False
    
    # Unfreeze the last convolutional layer (layer4)
    for param in model.layer4.parameters():
        param.requires_grad = True
    
    # Replace and unfreeze the final fully connected layer (classifier)
    num_ftrs = model.fc.in_features
    model.fc = nn.Linear(num_ftrs, num_classes)
    
    return model

model = create_model(num_classes)
model = model.to(device)

# Print which parameters will be trained
print("Layers being trained:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"\t{name}")

# Loss and optimizer
criterion = nn.CrossEntropyLoss()
# Using a slightly lower learning rate for the convolutional layers
optimizer = optim.Adam([
    {'params': model.layer4.parameters(), 'lr': 5e-5},  # Lower learning rate for conv layers
    {'params': model.fc.parameters(), 'lr': 1e-4}       # Higher learning rate for fc layer
])
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.3, patience=3, verbose=True)

# Training function with early stopping logic
def train_model(model, criterion, optimizer, scheduler, num_epochs=20, patience=5):
    best_model_wts = model.state_dict()
    best_acc = 0.0
    epochs_no_improve = 0
    
    history = {
        'train_loss': [], 'train_acc': [],
        'val_loss': [], 'val_acc': []
    }
    
    for epoch in range(num_epochs):
        print(f'Epoch {epoch+1}/{num_epochs}')
        print('-' * 10)
        
        for phase in ['train', 'val']:
            model.train() if phase == 'train' else model.eval()
            running_loss = 0.0
            running_corrects = 0
            
            for inputs, labels in tqdm(dataloaders[phase], desc=f"{phase} batches"):
                inputs = inputs.to(device)
                labels = labels.to(device)
                
                optimizer.zero_grad()
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()
                
                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)
            
            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]
            
            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')
            
            if phase == 'train':
                history['train_loss'].append(epoch_loss)
                history['train_acc'].append(epoch_acc.item())
            else:
                history['val_loss'].append(epoch_loss)
                history['val_acc'].append(epoch_acc.item())
                scheduler.step(epoch_loss)
                if epoch_acc > best_acc:
                    best_acc = epoch_acc
                    best_model_wts = model.state_dict().copy()
                    epochs_no_improve = 0
                    torch.save({
                        'epoch': epoch,
                        'model_state_dict': best_model_wts,
                        'optimizer_state_dict': optimizer.state_dict(),
                        'accuracy': best_acc,
                    }, os.path.join(model_logs_dir, 'best_model.pth'))
                else:
                    epochs_no_improve += 1
                    if epochs_no_improve >= patience:
                        print("Early stopping triggered.")
                        model.load_state_dict(best_model_wts)
                        return model, history
        
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'history': history,
        }, os.path.join(model_logs_dir, f'checkpoint_epoch_{epoch+1}.pth'))
        print()
    
    model.load_state_dict(best_model_wts)
    return model, history

print("Starting model training...")
model, history = train_model(model, criterion, optimizer, scheduler, num_epochs=20, patience=5)
print("Training complete!")

# Plot training history
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history['train_loss'], label='Train Loss')
plt.plot(history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Loss Over Time')

plt.subplot(1, 2, 2)
plt.plot(history['train_acc'], label='Train Accuracy')
plt.plot(history['val_acc'], label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.title('Accuracy Over Time')
plt.tight_layout()
plt.savefig(os.path.join(model_logs_dir, 'training_history.png'))
plt.show()

# Evaluate on test set
def evaluate_model(model, dataloader):
    model.eval()
    all_preds = []
    all_labels = []
    
    with torch.no_grad():
        for inputs, labels in tqdm(dataloader, desc='Evaluating'):
            inputs = inputs.to(device)
            labels = labels.to(device)
            
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    
    return all_preds, all_labels

print("Evaluating on test set...")
y_pred, y_true = evaluate_model(model, dataloaders['test'])
accuracy = np.mean(np.array(y_pred) == np.array(y_true))
print(f"Test Accuracy: {accuracy:.4f}")

# Detailed classification metrics
print("\nClassification Report:")
# Need to get the actual class names
class_names_list = datasets['train'].dataset.classes
print(classification_report(y_true, y_pred, target_names=class_names_list))

# Compute F1, Precision, Recall
f1 = f1_score(y_true, y_pred, average='macro')
precision = precision_score(y_true, y_pred, average='macro')
recall = recall_score(y_true, y_pred, average='macro')

print(f"Macro F1 Score: {f1:.4f}")
print(f"Macro Precision: {precision:.4f}")
print(f"Macro Recall: {recall:.4f}")

# Confusion Matrix
conf_mat = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(conf_mat, annot=True, fmt='d', cmap='Blues', 
            xticklabels=class_names_list, 
            yticklabels=class_names_list)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.tight_layout()
plt.savefig(os.path.join(model_logs_dir, 'confusion_matrix.png'))
plt.show()

# Save model and metrics
model_info = {
    'model_state_dict': model.state_dict(),
    'class_names': class_names_list,
    'metrics': {
        'accuracy': accuracy,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }
}

torch.save(model_info, os.path.join(model_logs_dir, 'final_model.pth'))
print(f"Model saved to {os.path.join(model_logs_dir, 'final_model.pth')}")

# Function to make predictions on new images
def predict_image(image_path, model, transform=val_test_transform):
    from PIL import Image
    
    model.eval()
    img = Image.open(image_path).convert('RGB')
    img_tensor = transform(img).unsqueeze(0).to(device)
    
    with torch.no_grad():
        outputs = model(img_tensor)
        _, preds = torch.max(outputs, 1)
        
    return class_names_list[preds[0]]

# Example usage:
# predicted_class = predict_image('path/to/your/image.jpg', model)
# print(f"Predicted class: {predicted_class}")
```

```python
# Imports remain the same
from sklearn.metrics import confusion_matrix, classification_report, f1_score, precision_score, recall_score
import os
import torch
import numpy as np
from torchvision import transforms, models
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader, Dataset
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from tqdm import tqdm
import seaborn as sns
import pandas as pd
from PIL import Image

# Set seeds
torch.manual_seed(42)
np.random.seed(42)
device = torch.device('cpu')
print(f"Using device: {device}")

# Directories
processed_data_dir = './processedData'
model_logs_dir = './modelLogs'
os.makedirs(model_logs_dir, exist_ok=True)

# Define class-specific transforms
light_transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize
    transforms.Pad(padding=4),      # Padding
    transforms.RandomHorizontalFlip(),  # Flipping
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])  # Normalize
])

heavy_transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize
    transforms.Pad(padding=4),      # Padding
    transforms.RandomHorizontalFlip(),  # Flipping
    transforms.RandomVerticalFlip(),    # Flipping
    transforms.RandomRotation(15),      # Rotation
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3),  # Color jitter
    transforms.RandomAffine(degrees=0, translate=(0.1, 0.1)),  # Affine transformation
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])  # Normalize
])

# Validation and test transforms (simple)
val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize
    transforms.Pad(padding=4),      # Padding
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])  # Normalize
])

# Custom dataset for class-specific transforms
class SkinLesionDataset(Dataset):
    def __init__(self, root_dir, split, light_classes=None):
        self.dataset = ImageFolder(os.path.join(root_dir, split))
        self.split = split
        # If no light_classes are specified, default to "melanoma" and "melanocytic nevi"
        self.light_classes = light_classes or ["melanoma", "melanocytic nevi"]
        self.light_class_indices = [self.dataset.class_to_idx[cls] for cls in self.light_classes if cls in self.dataset.class_to_idx]
        
        print(f"Light augmentation applied to classes: {self.light_classes}")
        print(f"Light augmentation class indices: {self.light_class_indices}")
        
    def __len__(self):
        return len(self.dataset)
    
    def __getitem__(self, idx):
        image_path, label = self.dataset.samples[idx]
        image = Image.open(image_path).convert('RGB')
        
        # Apply transforms based on class and split
        if self.split == 'train':
            if label in self.light_class_indices:
                # Light augmentation for melanoma and melanocytic nevi
                transform = light_transform
            else:
                # Heavy augmentation for other classes
                transform = heavy_transform
        else:
            # Simple transforms for validation and test
            transform = val_test_transform
            
        image = transform(image)
        return image, label

# Load datasets with class-specific augmentation
print("Loading datasets with class-specific augmentation...")
datasets = {
    split: SkinLesionDataset(processed_data_dir, split, ["melanoma", "melanocytic nevi"])
    for split in ['train', 'val', 'test']
}

# Dataloaders
dataloaders = {
    split: DataLoader(datasets[split], batch_size=16, shuffle=(split == 'train'), num_workers=0)
    for split in ['train', 'val', 'test']
}

# Get class names (need to use the underlying ImageFolder dataset)
class_names = datasets['train'].dataset.classes
num_classes = len(class_names)
dataset_sizes = {split: len(datasets[split]) for split in ['train', 'val', 'test']}

print(f"Number of classes: {num_classes}")
print(f"Class names: {class_names}")
for split in ['train', 'val', 'test']:
    print(f"{split.capitalize()} set size: {dataset_sizes[split]}")

# Display sample images
def imshow(inp, title=None):
    inp = inp.cpu().numpy().transpose((1, 2, 0))
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    inp = std * inp + mean
    inp = np.clip(inp, 0, 1)
    plt.figure(figsize=(15, 15))
    plt.imshow(inp)
    if title is not None:
        plt.title(title)
    plt.pause(0.001)

try:
    inputs, classes = next(iter(dataloaders['train']))
    out = make_grid(inputs[:4])
    # Need to get class names mapped to indices
    class_labels = [datasets['train'].dataset.classes[x] for x in classes[:4]]
    imshow(out, title=str(class_labels))
except Exception as e:
    print(f"Error displaying sample images: {e}")

# Custom ResNet model with additional hidden layers and dropout
class ResNetWithMLPHead(nn.Module):
    def __init__(self, num_classes, dropout_rate=0.5):
        super(ResNetWithMLPHead, self).__init__()
        # Load pretrained ResNet18
        resnet = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
        
        # Remove the original fully connected layer
        self.features = nn.Sequential(*list(resnet.children())[:-1])
        
        # Get the output dimension of the last convolutional layer
        self.feature_dim = resnet.fc.in_features
        
        # Add an MLP head with hidden layers and dropout
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(self.feature_dim, 512),
            nn.BatchNorm1d(512),
            nn.ReLU(),
            nn.Dropout(dropout_rate),
            nn.Linear(512, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(dropout_rate),
            nn.Linear(256, num_classes)
        )
        
        # Freeze all layers initially
        for param in self.features.parameters():
            param.requires_grad = False
        
        # Unfreeze the last three layers (layer3, layer4, and the classifier)
        for param in self.features[6].parameters():  # layer3
            param.requires_grad = True
        for param in self.features[7].parameters():  # layer4
            param.requires_grad = True
    
    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

# Create model
model = ResNetWithMLPHead(num_classes, dropout_rate=0.5)
model = model.to(device)

# Print which parameters will be trained
print("Layers being trained:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"\t{name}")

# Calculate total trainable parameters
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total trainable parameters: {trainable_params:,}")

# Loss and optimizer
criterion = nn.CrossEntropyLoss()
# Using different learning rates for different layer groups
optimizer = optim.Adam([
    {'params': model.features[6].parameters(), 'lr': 3e-5},  # layer3
    {'params': model.features[7].parameters(), 'lr': 5e-5},  # layer4
    {'params': model.classifier.parameters(), 'lr': 1e-4}    # MLP classifier
])
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.3, patience=3, verbose=True)

# Training function with early stopping logic
def train_model(model, criterion, optimizer, scheduler, num_epochs=20, patience=5):
    best_model_wts = model.state_dict()
    best_acc = 0.0
    epochs_no_improve = 0
    
    history = {
        'train_loss': [], 'train_acc': [],
        'val_loss': [], 'val_acc': []
    }
    
    for epoch in range(num_epochs):
        print(f'Epoch {epoch+1}/{num_epochs}')
        print('-' * 10)
        
        for phase in ['train', 'val']:
            model.train() if phase == 'train' else model.eval()
            running_loss = 0.0
            running_corrects = 0
            
            for inputs, labels in tqdm(dataloaders[phase], desc=f"{phase} batches"):
                inputs = inputs.to(device)
                labels = labels.to(device)
                
                optimizer.zero_grad()
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)
                    if phase == 'train':
                        loss.backward()
                        optimizer.step()
                
                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)
            
            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]
            
            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')
            
            if phase == 'train':
                history['train_loss'].append(epoch_loss)
                history['train_acc'].append(epoch_acc.item())
            else:
                history['val_loss'].append(epoch_loss)
                history['val_acc'].append(epoch_acc.item())
                scheduler.step(epoch_loss)
                if epoch_acc > best_acc:
                    best_acc = epoch_acc
                    best_model_wts = model.state_dict().copy()
                    epochs_no_improve = 0
                    torch.save({
                        'epoch': epoch,
                        'model_state_dict': best_model_wts,
                        'optimizer_state_dict': optimizer.state_dict(),
                        'accuracy': best_acc,
                    }, os.path.join(model_logs_dir, 'best_model.pth'))
                else:
                    epochs_no_improve += 1
                    if epochs_no_improve >= patience:
                        print("Early stopping triggered.")
                        model.load_state_dict(best_model_wts)
                        return model, history
        
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'history': history,
        }, os.path.join(model_logs_dir, f'checkpoint_epoch_{epoch+1}.pth'))
        print()
    
    model.load_state_dict(best_model_wts)
    return model, history

print("Starting model training...")
model, history = train_model(model, criterion, optimizer, scheduler, num_epochs=20, patience=5)
print("Training complete!")

# Plot training history
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history['train_loss'], label='Train Loss')
plt.plot(history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Loss Over Time')

plt.subplot(1, 2, 2)
plt.plot(history['train_acc'], label='Train Accuracy')
plt.plot(history['val_acc'], label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.title('Accuracy Over Time')
plt.tight_layout()
plt.savefig(os.path.join(model_logs_dir, 'training_history.png'))
plt.show()

# Evaluate on test set
def evaluate_model(model, dataloader):
    model.eval()
    all_preds = []
    all_labels = []
    
    with torch.no_grad():
        for inputs, labels in tqdm(dataloader, desc='Evaluating'):
            inputs = inputs.to(device)
            labels = labels.to(device)
            
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    
    return all_preds, all_labels

print("Evaluating on test set...")
y_pred, y_true = evaluate_model(model, dataloaders['test'])
accuracy = np.mean(np.array(y_pred) == np.array(y_true))
print(f"Test Accuracy: {accuracy:.4f}")

# Detailed classification metrics
print("\nClassification Report:")
# Need to get the actual class names
class_names_list = datasets['train'].dataset.classes
print(classification_report(y_true, y_pred, target_names=class_names_list))

# Compute F1, Precision, Recall
f1 = f1_score(y_true, y_pred, average='macro')
precision = precision_score(y_true, y_pred, average='macro')
recall = recall_score(y_true, y_pred, average='macro')

print(f"Macro F1 Score: {f1:.4f}")
print(f"Macro Precision: {precision:.4f}")
print(f"Macro Recall: {recall:.4f}")

# Confusion Matrix
conf_mat = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(conf_mat, annot=True, fmt='d', cmap='Blues', 
            xticklabels=class_names_list, 
            yticklabels=class_names_list)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.tight_layout()
plt.savefig(os.path.join(model_logs_dir, 'confusion_matrix.png'))
plt.show()

# Save model and metrics
model_info = {
    'model_state_dict': model.state_dict(),
    'class_names': class_names_list,
    'metrics': {
        'accuracy': accuracy,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }
}

torch.save(model_info, os.path.join(model_logs_dir, 'FINALMODEL.pth'))
print(f"Model saved to {os.path.join(model_logs_dir, 'FINALMODEL.pth')}")


# Function to make predictions on new images
def predict_image(image_path, model, transform=val_test_transform):
    from PIL import Image
    
    model.eval()
    img = Image.open(image_path).convert('RGB')
    img_tensor = transform(img).unsqueeze(0).to(device)
    
    with torch.no_grad():
        outputs = model(img_tensor)
        _, preds = torch.max(outputs, 1)
        
    return class_names_list[preds[0]]

# Example usage:
# predicted_class = predict_image('path/to/your/image.jpg', model)
# print(f"Predicted class: {predicted_class}")
```
